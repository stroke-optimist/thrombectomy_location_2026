```python
p = './data/pivoted_time.csv'
lsoa_unit_times = pd.read_csv(p, index_col=0)
```

```python
s_nearest_lsoa_units = pd.Series()
s_nearest_lsoa_units_mt = pd.Series()
lsoa_with_multiple_nearest = []
lsoa_with_multiple_nearest_mt = []

for lsoa in lsoa_unit_times.index:
    s = lsoa_unit_times.loc[lsoa]
    t = s.min()
    units_nearest = s[s == t].index.tolist()
    s_nearest_lsoa_units[lsoa] = units_nearest
    if len(units_nearest) > 1:
        lsoa_with_multiple_nearest.append(lsoa)

    s = lsoa_unit_times.loc[lsoa][mt_units]
    units_nearest = s[s == t].index.tolist()
    s_nearest_lsoa_units_mt[lsoa] = units_nearest

    if len(units_nearest) > 1:
        lsoa_with_multiple_nearest_mt.append(lsoa)
```
