## Code setup

```python
import os
import pandas as pd

from classes.calc_scenario import load_geo_cohort_data
```

```python
dir_output = 'output'
# Scenario results:
file_results_comparison = 'results_full_comparison_taunworc_base'  # _bias.csv
file_geo_outcomes = 'results_geo_outcomes_'  # base, base_bias.csv 
file_results_full = 'results_full_'  # base, base_bias.csv 
file_unit_benefit = 'benefit_of_each_team.csv'
```

## Load data

```python
col_admissions = 'Admissions_2025_Est'
```

```python
postcode_taun = 'TA15DA'
postcode_worc = 'WR51DD'
```

```python
dfs_england_comp_file_dict = {
    'nobias': f'{file_results_comparison}.csv',
    'bias10': f'{file_results_comparison}_bias10min.csv',
    'bias15': f'{file_results_comparison}_bias15min.csv',
}
dfs_england_comp_dict = {}
for label, file_name in dfs_england_comp_file_dict.items():
    p = os.path.join(dir_output, file_name)
    df = pd.read_csv(p, index_col=0)
    dfs_england_comp_dict[label] = df
```

```python
dfs_england_comp_dict['nobias'].columns
```

```python
scenario_file_strings = ['base', 'taunworc']
scenario_file_strings = [f'{s}{b}' for s in scenario_file_strings
                         for b in ['', '_bias10min', '_bias15min']]

dfs_geo_dict = {}
for s in scenario_file_strings:
    # Load in data:
    p = os.path.join(dir_output, file_geo_outcomes + s + '.csv')
    p2 = os.path.join(dir_output, file_results_full + s + '.csv')
    df = load_geo_cohort_data(p, p2)

    dfs_geo_dict[s] = df
```

## Maximum travel times

```python
cols = ['drip_ship_mt_travel_time_with_delay_base',
        'drip_ship_mt_travel_time_with_delay_taunworc']

for label, df in dfs_england_comp_dict.items():
    print(label)
    display(df[cols].describe().loc[['min', 'max']])
```

## Direct admissions to MT unit

```python
for label, df in dfs_england_comp_dict.items():
    print(label)
    for c in ['transfer_required_base', 'transfer_required_taunworc']:
        v = ((df[c] * df[col_admissions]).sum() /
             df[col_admissions].sum())
        print(f'{100 - 100*v:.1f}% admissions direct to MT unit.')
    print('')
```

## Average proportion with good outcome

In England

```python
cols_dict = {
    'lvo_no_treatment_mrs_0-2': 'No treatment',
    'lvo_drip_ship_ivt_mrs_0-2': 'Drip-and-ship IVT',
    'lvo_drip_ship_ivt_mt_mrs_0-2': 'Drip-and-ship IVT & MT',
    'lvo_mothership_ivt_mrs_0-2': 'Mothership IVT',
    'lvo_mothership_ivt_mt_mrs_0-2': 'Mothership IVT & MT',
    'lvo_drip_ship_ivt_mt_mrs_0-2_improvement_over_no_treatment': ' Drip-and-ship improvement',
    'lvo_drip_ship_ivt_mt_mrs_0-2_improvement_over_no_treatment_transferloss0!67': ' Drip-and-ship improvement (67% transfer)'
}
cols_tl_dict = {}
for k, v in cols_dict.items():
    cols_tl_dict[f'{k}_transferloss0!67'] = v

df_lvo_outcomes = pd.DataFrame()

for label, df in dfs_geo_dict.items():
    v = ((df[list(cols_dict.keys())].mul(df[col_admissions], axis='rows')).sum() /
          df[col_admissions].sum())
    df_lvo_outcomes[label] = v

df_lvo_outcomes = df_lvo_outcomes.rename(index=cols_dict)
```

```python
print('Percentage of LVO patients with outcome mRS 0-2:')
display((100.0*df_lvo_outcomes).round(1))
```

## Range of proportion with good outcome

```python
cols_dict = {
    'lvo_no_treatment_mrs_0-2': 'No treatment',
    'lvo_drip_ship_ivt_mrs_0-2': 'Drip-and-ship IVT',
    'lvo_drip_ship_ivt_mt_mrs_0-2': 'Drip-and-ship IVT & MT',
    'lvo_mothership_ivt_mrs_0-2': 'Mothership IVT',
    'lvo_mothership_ivt_mt_mrs_0-2': 'Mothership IVT & MT',
    'lvo_drip_ship_ivt_mt_mrs_0-2_improvement_over_no_treatment': ' Drip-and-ship improvement'
}

cols_stats = ['min', 'max']
dfs = {}

for scen, df in dfs_geo_dict.items():
    v = df[cols_dict.keys()].describe().loc[cols_stats]
    v = v.rename(columns=cols_dict)
    dfs[scen] = v
    
df_lvo_props = pd.concat(dfs.values(), keys=dfs.keys(), axis='rows').transpose()
# Sort columns so stats are grouped instead of scenario names:
df_lvo_props = df_lvo_props[df_lvo_props.columns.sortlevel(1)[0]]
```

Show full results:

```python
display(df_lvo_props[df_lvo_props.columns.sortlevel(1)[0]])
```

Show just one stat at once:

```python
for s in cols_stats:
    print(s)
    display((100.0 * df_lvo_props.xs(s, axis='columns', level=1, drop_level=True)).round(1))
```
