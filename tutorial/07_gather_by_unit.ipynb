Take the results by combination of stroke units and combine rows that belong to the same unit.

```python
dir_output = os.path.join('..', 'output')
# Scenario results:
file_comparison_unit_combo = 'comparison_by_unit_combo_base_taunworc'
# Output:
file_comparison_nearest_unit = 'comparison_by_nearest_unit_base_taunworc'
file_comparison_transfer_unit = 'comparison_by_transfer_unit_base_taunworc'
```

```python
strings_tl = ['', '_transferloss0!67']  # column names for transfer rates
scenarios = ['base', 'taunworc']
s_bias = ''
```

Run this example for this scenario and unit type:

```python
s = 'base'
col_unit = f'transfer_unit_{s}'  # f'nearest_ivt_unit_{s}'
```

## Load data

Load in the outcomes and admissions data by combination of units in the base and Taunton&Worcester scenarios:

```python
p = os.path.join(dir_output, file_comparison_unit_combo + s_bias + '.csv')
comparison_by_unit_combo = pd.read_csv(p, index_col=0)
```

## Setup

Gather columns present in the main results dataframes.

Gather columns by type and without tailing scenario names (`_base`, `_taunworc`):

```python
# Set up columns:
cols_scen = [c for c in comparison_by_unit_combo.columns if s in c]
cols_adweight = [c for c in cols_scen if 'weighted' in c]
cols_time_here = [c for c in cols_scen if 'drip_ship_mt_' in c]
cols_ad = [c for c in cols_scen if
           ((('dmissions' in c) | ('transfers_mt' in c)) &
            (c not in cols_adweight))]

col_admissions = 'Admissions_2025_Est'
col_ad = f'{col_admissions}_{s}'
```

Decide which units to show in this example.

Pick out units that appear more than once for this unit type and scenario:

```python
s_counts = comparison_by_unit_combo[col_unit].value_counts()
units_multiple = sorted(s_counts[s_counts > 1].index.tolist())

print(units_multiple)
```

Instead just show Bristol:

```python
units_to_show = ['BS105NB']  # Bristol
```

```python
units_to_show
```

Show combinations of units including the ones to be shown:

```python
m = comparison_by_unit_combo[col_unit].isin(units_to_show)
cols = [f'nearest_ivt_unit_{s}', f'Hospital_City_{s}',
        f'transfer_unit_{s}', f'Hospital_City_transfer_unit_{s}']
display(comparison_by_unit_combo.loc[m, cols].sort_values(col_unit).head(10))
```

Define this function for displaying results in this notebook:

```python
def display_df_in_unit(df, col_unit, units_to_show):
    display(
        df[df[col_unit].isin(units_to_show)]
        .sort_values(col_unit)
        .head(10)
        )
```

## Group results by unit

### Stroke unit info

Gather the SSNAP name and hospital city for this type of unit.

```python
# Stroke unit info:
if 'transfer' in col_unit:
    cols_this_unit = [f'SSNAP name_transfer_unit_{s}',
                      f'Hospital_City_transfer_unit_{s}']
else:
    cols_this_unit = [f'SSNAP name_{s}', f'Hospital_City_{s}']

df_units = (comparison_by_unit_combo[[col_unit] + cols_this_unit]
    .drop_duplicates(ignore_index=True))
df_units = df_units.set_index(col_unit)
```

```python
display(df_units.head(3))
```

### Admissions

Admissions can just be summed.

Pick out and display the starting data across multiple rows:

```python
df_ad = comparison_by_unit_combo[[col_unit] + cols_ad].copy()

display_df_in_unit(df_ad, col_unit, units_to_show)
```

Sum the values into one row and show the results:

```python
df_ad = df_ad.groupby(col_unit).sum()

display(df_ad.loc[units_to_show].head(10))
```

### Admissions-weighted outcomes

Cannot just sum the outcomes, but have to factor in how many admissions represent each outcome.

Note: the starting columns names say "admissions-weighted". This refers to them being an average across all patients in this row. The outcomes for proportion of mRS 0-2 are all between 0 and 1.

```python
df_adweight = comparison_by_unit_combo[
    [col_unit, col_ad] + cols_adweight].copy()

display_df_in_unit(df_adweight, col_unit, units_to_show)
```

First multiply outcomes by the number of admissions. This tells us how many patients receive an outcome of mRS 0-2.

```python
df_adweight[cols_adweight] = (
    df_adweight[cols_adweight]
        .multiply(df_adweight[col_ad], axis='rows')
)

display_df_in_unit(df_adweight, col_unit, units_to_show)
```

Then sum all of the rows into one:

```python
cols = [col_unit, col_ad] + cols_adweight
df_adweight = df_adweight[cols].groupby(col_unit).sum()

display(df_adweight.loc[units_to_show].head(10))
```

Divide the summed outcomes by the direct admissions to each unit to convert total number of good outcomes into proportion of good outcomes.

```python
df_adweight[cols_adweight] = (
    df_adweight[cols_adweight]
        .div(df_adweight[col_ad], axis='rows')
)

display(df_adweight.loc[units_to_show].head(10))
```

Prepare for combining this dataframe with others: round the values and drop the admissions column, which would otherwise appear multiple times.

```python
df_adweight = df_adweight.drop(col_ad, axis='columns')
df_adweight = df_adweight.round(5)
```

### Treatment times

Same principle as the admissions-weighted outcomes.

Start with average time to treatment per unit combo:

```python
df_times = comparison_by_unit_combo[
    [col_unit, col_ad] + cols_time_here].copy()

display_df_in_unit(df_times, col_unit, units_to_show)
```

Convert average time per patient to total time for all patients per unit combo:

```python
df_times[cols_time_here] = df_times[cols_time_here].multiply(
    df_times[col_ad], axis='rows')

display_df_in_unit(df_times, col_unit, units_to_show)
```

Sum unit combos to find the total time for just the one row per unit:

```python
cols = [col_unit, col_ad] + cols_time_here
df_times = df_times[cols].groupby(col_unit).sum()

display(df_times.loc[units_to_show].head(10))
```

Divide total times by the total number of patients to each unit to find the average time for a patient going to that unit:

```python
df_times[cols_time_here] = df_times[cols_time_here].div(
    df_times[col_ad], axis='rows')

display(df_times.loc[units_to_show].head(10))
```

Prepare for combining this dataframe with others: round the values and drop the admissions column, which would otherwise appear multiple times.

```python
df_times = df_times.drop(col_ad, axis='columns')
df_times = df_times.round(1)
```

### Gather results by unit

```python
df_here = pd.concat((df_units, df_ad, df_adweight, df_times),
                    axis='columns')
```

```python
display(df_here.loc[units_to_show].head(10))
```
